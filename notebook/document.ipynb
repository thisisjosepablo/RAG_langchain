```python
### TextLoader
from pathlib import Path
def load_text_file(file_path:str, encoding: str="utf-8") -> list[Document]:
    path = Path(file_path)
    content = path.read_text(encoding=encoding)

    return [
        Document(
            page_content = content,
            metadata = {
                "source": str(path.resolve())
            })
        ]


docs = load_text_file ("../data/text_files/guardiacivil_calcetines.txt")
print(docs[0].page_content)
print(docs[0].metadata)

```

```python
### Directory Loader
# It allows to load a bunch of files which are inside a folder


def directory_loader_txt(dir_path: str, encoding: str = "utf-8") -> list[Document]:
    dir_path = Path(dir_path)
    documents = []
    for file in dir_path.iterdir():
        if file.suffix == ".txt":
            file_path = str(file.resolve())
            documents.extend(load_text_file(file_path))
    
    return documents

directory_loader_txt("../data/text_files")
```

```python
import openpyxl

path = "../data/excel/patients.xlsx"

def excel_loader(path: str, header_q: bool = True) -> list[Document]:

    path = Path(path)
    wb = openpyxl.load_workbook(path)
    ws = wb.active

    header = []
    if header_q:
        header = [ws.cell(row = 1, column = i).value for i in range(1, ws.max_column+1)]
    
    content = ""
    for i in range(2, ws.max_row+1):
        row = ""
        for j in range(1, ws.max_column+1):
            row += f"{str(ws.cell(row=i, column=j).value)},"
        row += '\n'
        content += row

    return [
        Document(
            page_content = content,
            metadata = {
                "header" : header,
                "source" : str(path.resolve())
            }
        )
    ]


print(excel_loader(path))
```
